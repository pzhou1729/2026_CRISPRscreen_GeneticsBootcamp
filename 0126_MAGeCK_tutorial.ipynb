Let's make some simple volcano plots for each of these datasets. Note: a two-sided test provides evidence that LFC is significantly different from 0, so the column titled p.twosided should be used for the sgRNA volcano plot.

```python
#sgRNA volcano plot 

#prep
df = sgrna_summary.copy()

df["LFC"] = pd.to_numeric(df["LFC"], errors="coerce")
df["p"] = pd.to_numeric(df["p.twosided"], errors="coerce")
df = df.dropna(subset=["LFC", "p"])
df = df[df["p"] > 0]

df["neglog10_p"] = -np.log10(df["p"])

#gene masks
ldlr_mask = df["Gene"] == "LDLR"
atp6v1g2_mask = df["Gene"] == "ATP6V1G2"

#base plot
plt.figure(figsize=(6, 5))

plt.scatter(
    df["LFC"],
    df["neglog10_p"],
    s=6,
    alpha=0.35
)

#circle LDLR
plt.scatter(
    df.loc[ldlr_mask, "LFC"],
    df.loc[ldlr_mask, "neglog10_p"],
    s=90,
    facecolors="none",
    linewidths=1.8
)

#circle ATP6V1G2
plt.scatter(
    df.loc[atp6v1g2_mask, "LFC"],
    df.loc[atp6v1g2_mask, "neglog10_p"],
    s=90,
    facecolors="none",
    linewidths=1.8
)

#label points
for _, row in df.loc[ldlr_mask].iterrows():
    plt.text(
        row["LFC"],
        row["neglog10_p"],
        "LDLR",
        fontsize=8,
        ha="center",
        va="bottom"
    )

for _, row in df.loc[atp6v1g2_mask].iterrows():
    plt.text(
        row["LFC"],
        row["neglog10_p"],
        "ATP6V1G2",
        fontsize=8,
        ha="center",
        va="bottom"
    )

# --- thresholds ---
plt.axvline(1, linestyle="--")
plt.axvline(-1, linestyle="--")
plt.axhline(-np.log10(0.05), linestyle="--")

plt.xlabel("Log2 Fold Change (LFC)")
plt.ylabel("-log10(p-value)")
plt.title("Serum: sgRNA Volcano Plot")

plt.tight_layout()
plt.show()
```

```python
df = gene_summary.copy()

# --- extract values ---
lfc = pd.to_numeric(df["neg|lfc"], errors="coerce")
p   = pd.to_numeric(df["neg|p-value"], errors="coerce")

mask = (p > 0) & lfc.notna() & p.notna()

x = lfc[mask]
y = -np.log10(p[mask])

# --- base plot ---
plt.figure(figsize=(6, 5))
plt.scatter(x, y, s=10, alpha=0.5)

# --- highlight MYLIP ---
mylip_mask = mask & (df["id"] == "MYLIP")

plt.scatter(
    lfc[mylip_mask],
    -np.log10(p[mylip_mask]),
    s=120,
    facecolors="none",
    linewidths=2
)

# --- label MYLIP ---
for xi, yi in zip(lfc[mylip_mask], -np.log10(p[mylip_mask])):
    plt.text(
        xi,
        yi,
        "MYLIP",
        fontsize=9,
        ha="center",
        va="bottom"
    )

# --- thresholds ---
plt.axvline(1, linestyle="--")
plt.axvline(-1, linestyle="--")
plt.axhline(-np.log10(0.05), linestyle="--")

plt.xlabel("neg|lfc")
plt.ylabel("-log10(neg|p-value)")
plt.title("Gene-level Volcano Plot (negative selection)")

plt.tight_layout()
plt.show()
```

```python
df = gene_summary.copy()

lfc = pd.to_numeric(df["pos|lfc"], errors="coerce")
p   = pd.to_numeric(df["pos|p-value"], errors="coerce")

mask = (p > 0) & lfc.notna() & p.notna()

x = lfc[mask]
y = -np.log10(p[mask])

# --- base plot ---
plt.figure(figsize=(6, 5))
plt.scatter(x, y, s=10, alpha=0.5)

# --- highlight LDLR ---
ldlr_mask = mask & (df["id"] == "LDLR")

plt.scatter(
    lfc[ldlr_mask],
    -np.log10(p[ldlr_mask]),
    s=120,
    facecolors="none",
    linewidths=2
)

# --- label LDLR ---
for xi, yi in zip(lfc[ldlr_mask], -np.log10(p[ldlr_mask])):
    plt.text(
        xi,
        yi,
        "LDLR",
        fontsize=9,
        ha="center",
        va="bottom"
    )

# --- thresholds ---
plt.axvline(1, linestyle="--")
plt.axvline(-1, linestyle="--")
plt.axhline(-np.log10(0.05), linestyle="--")

plt.xlabel("pos|lfc")
plt.ylabel("-log10(pos|p-value)")
plt.title("Serum: Gene-level Volcano Plot (positive selection)")

plt.tight_layout()
plt.show()
```
